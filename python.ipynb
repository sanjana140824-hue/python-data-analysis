ABC language was very simple to read and write, but it had some limits. During his Christmas holidays in 1989, Guido thought there should be a language that is both simple and powerful. And that is how Python started! He released the first version of Python in 1991.

Many people think that Python is named after the snake. But that is not true! Guido loved a British comedy show called **"Monty Python's Flying Circus"**. He wanted a name that was short, unique, and slightly mysterious, so he chose the name **Python**.

# **How Python Works: Compiled vs. Interpreted Languages**

To understand how Python works, we need to know how computers read code. Computers cannot understand human languages or even English-like code directly; they only understand **machine code** (0s and 1s).

There are two main ways to translate our code into machine code:

### **1. Compiled Languages (Compile-time)**
* **How it works:** In a compiled language (like C, C++, or Java), a special program called a **compiler** translates the entire source code into machine code all at once before the program runs.
* **Analogy:** It is like translating a whole book from English to Spanish and then giving the Spanish book to the reader.
* **Pros & Cons:** It runs very fast, but if you make a small change, you have to compile the whole program again.

### **2. Interpreted Languages**
* **How it works:** In an interpreted language (like Python), a program called an **interpreter** reads and runs the code line by line, translating it on the spot as the program is running.
* **Analogy:** It is like a live translator who translates a speech sentence by sentence as the speaker talks.
* **Pros & Cons:** It is much easier to write and test your code because you can run it instantly. However, it can run a bit slower than compiled languages.

### **What is Python?**
Python is an **interpreted language**. When you run a Python script, the Python interpreter reads your code line by line and executes it immediately!

# **Comments and Variables in Python**

Let's learn how to use comments and variables with a simple, practical example.

### **What is a Comment?**
A **comment** is a note written inside the code for humans. The computer completely ignores it when running the program.
* `#` is used for writing single-line comments.
* For multi-line comments, we can use multi-line strings with triple quotes (`'''` or `"""`).

### **What is a Variable?**
A **variable** is like a labeled container or box used to store data (like text or numbers) so we can use it later.
* **Example:**
  ```python
  a = 10
  b = 'sanjana'
  ```

### **Rules for Naming Variables**
To name a variable in Python, you must follow these simple rules:
1. **Must start with a letter or underscore (_)**: For example, `name` or `_name` is correct, but `1name` is incorrect.
2. **Can only contain letters, numbers, and underscores**: For example, `user_age1` is correct, but `user-age` or `user age` (with spaces) is incorrect.
3. **Case-sensitive**: `Age`, `age`, and `AGE` are three different variables.
4. **Cannot use Python keywords**: You cannot use special Python words like `print`, `if`, or `for` as variable names.

---

### **Variable Naming Conventions**
While the rules above are mandatory, Python developers also follow style guides (like PEP 8) to make code easier to read. These are called **naming conventions**:

* **snake_case**: Use lowercase letters and separate words with underscores. This is the standard way to write variable names in Python.
  * **Example:** `user_name`, `total_price`, `my_favorite_food`
* **camelCase**: The first word is lowercase, and each following word starts with a capital letter.
  * **Example:** `userName`, `totalPrice`
* **PascalCase**: Every word starts with a capital letter (mostly used for naming classes in Python).
  * **Example:** `UserName`, `TotalPrice`

### **2. String Indexing**
In Python, every character in a string is stored at a specific position, called an **index**. Think of indices like address numbers on houses in a street.

* **Indexing starts at `0`**: The very first character of a string is always at index `0`, the second is at index `1`, and so on.
* **Negative Indexing**: Python also lets you count backward from the end of the string. The last character is at index `-1`, the second-to-last is at `-2`, and so on.

#### **Visual Representation of the String `"PYTHON"`**

| Character | P | Y | T | H | O | N |
| :--- | :---: | :---: | :---: | :---: | :---: | :---: |
| **Positive Index (Left to Right)** | `0` | `1` | `2` | `3` | `4` | `5` |
| **Negative Index (Right to Left)** | `-6` | `-5` | `-4` | `-3` | `-2` | `-1` |

#### **Examples of Indexing:**
```python
word = "PYTHON"

# Accessing using positive index
print(word[0])  # Output: 'P'
print(word[3])  # Output: 'H'

# Accessing using negative index
print(word[-1]) # Output: 'N' (last character)
print(word[-4]) # Output: 'T'
```

## **2. Operators in Python**

Operators are special symbols used to perform calculations, assignments, or logical comparisons on variables and values.

Here are the primary types of operators in Python:

### **A. Arithmetic Operators** (For mathematical calculations)
* `+` : Addition -> `10 + 5 = 15`
* `-` : Subtraction -> `10 - 5 = 5`
* `*` : Multiplication -> `10 * 5 = 50`
* `/` : Division -> `10 / 4 = 2.5` (always returns a float)
* `//` : Floor Division -> `10 // 4 = 2` (rounds down to the nearest integer)
* `%` : Modulus -> `10 % 3 = 1` (returns the remainder of division)
* `**` : Exponentiation (Power) -> `2 ** 3 = 8` ($2^3$)

### **B. Assignment Operators** (To store values in variables)
* `=` : Assign value -> `x = 5`
* `+=` : Add and assign -> `x += 3` (equivalent to `x = x + 3`)
* `-=` : Subtract and assign -> `x -= 2` (equivalent to `x = x - 2`)

### **C. Comparison Operators** (To compare values - always returns `True` or `False`)
* `==` : Equal to -> `5 == 5` is `True`
* `!=` : Not equal to -> `5 != 3` is `True`
* `>` : Greater than -> `5 > 3` is `True`
* `<` : Less than -> `5 < 3` is `False`
* `>=` : Greater than or equal to
* `<=` : Less than or equal to

### **D. Logical Operators** (To combine multiple conditions)
* `and` : Returns `True` if both conditions are true.
* `or` : Returns `True` if at least one of the conditions is true.
* `not` : Reverses the logical state (turns `True` to `False` and vice versa).

````python
# --- Arithmetic Operators Demo ---
a = 10
b = 3
print("Addition:", a + b)
print("Division:", a / b)
print("Floor Division (no decimal):", a // b)
print("Remainder (Modulus):", a % b)
print("Power (10 to the power 3):", a ** b)

# --- Comparison Operators Demo ---
x = 10
y = 20
print("Is x equal to y?:", x == y)
print("Is x less than y?:", x < y)

# --- Logical Operators Demo ---
is_student = True
has_id_card = False
print("Can enter library (both true)?:", is_student and has_id_card)
print("Can enter library (either one true)?:", is_student or has_id_card)
````

## **3. If-Else & Conditional Statements**

Just like in real life, we need our programs to make decisions based on conditions. For this, we use **Conditional Statements**.

Python has 3 key decision-making keywords:
1. **`if`**: The block of code inside the `if` block executes only if the condition is `True`.
2. **`elif`** (Else If): If the previous condition is false, we use `elif` to check another condition.
3. **`else`**: If none of the conditions above are true, the `else` block runs automatically.

### **Indentation (Spacing) Rule:**
Unlike other programming languages that use curly brackets `{}` to define code blocks, Python uses **indentation** (typically 4 spaces or 1 tab). All indented code after a colon (`:`) belongs to that specific condition block.

````python
# Example: Checking if a person can vote
age = 19

if age >= 18:
    print("You are eligible to vote!")
else:
    print("Sorry, you are too young to vote.")

# Example 2: Grading System using elif
marks = 85

if marks >= 90:
    print("Grade: A+")
elif marks >= 80:
    print("Grade: A")
elif marks >= 70:
    print("Grade: B")
else:
    print("Grade: C/Pass")
````

## **4. Loops Concept in Python**

When we need to execute a block of code multiple times, we use **Loops**. This prevents us from writing the same code over and over again.

Python has two main types of loops:

### **A. `for` Loop**
We use a `for` loop when we know in advance how many times we want to repeat a block of code. It iterates over a sequence (such as a list, string, or a range of numbers).
* **`range(start, stop)` function:** Generates a sequence of numbers from `start` up to (but not including) `stop`. For example, `range(1, 6)` generates `1, 2, 3, 4, 5`.

### **B. `while` Loop**
We use a `while` loop when we want to repeat code as long as a certain condition remains `True`. The loop stops when the condition becomes `False`.

````python
# --- 1. Demo of 'for' loop (Printing numbers 1 to 5) ---
print("Running for loop:")
for i in range(1, 6):
    print("Number:", i)

# --- 2. Demo of 'while' loop (Counting down from 5 to 1) ---
print("\nRunning while loop:")
count = 5
while count > 0:
    print("Count:", count)
    count -= 1  # count ko kam kar rahe hain taaki loop kabhi na kabhi ruke
````

## **5. While Loop Practice & Mini Game**

To solidify our understanding of `while` loops, let's create a simple Number Guessing Game. In this game, the program will repeatedly ask the user to guess a number until they get it correct.

````python
# Simple Number Guessing Game
secret_number = 7
guess = 0

print("--- Welcome to the Mini Guessing Game! ---")
print("Guess a number between 1 and 10!")

while guess != secret_number:
    guess = int(input("Enter your guess: "))

    if guess < secret_number:
        print("Too low! Try again.")
    elif guess > secret_number:
        print("Too high! Try again.")
    else:
        print("Congratulations! You guessed the right number!")
````

## **7. Data Structure Overview**

Data Structures are ways to organize, manage, and store data in a computer so that we can access and modify it efficiently. Python has 4 main built-in data structures:
1. **List**: Ordered, changeable, and allows duplicate items. `[1, 2, 2]`
2. **Tuple**: Ordered, unchangeable, and allows duplicate items. `(1, 2, 2)`
3. **Set**: Unordered, changeable, and does not allow duplicate items. `{1, 2, 3}`
4. **Dictionary**: Key-Value pairs, changeable, and does not allow duplicate keys. `{"name": "Sanjana"}`

### **8. Lists in Python**

A **List** is an ordered, changeable (mutable) collection of items that allows duplicate values. It is written using square brackets `[]`.

#### **Key Attributes & Characteristics:**
1. **Ordered:** Items have a defined order that will not change unless explicitly modified.
2. **Mutable (Changeable):** You can add, remove, and modify items after creation.
3. **Allows Duplicates:** You can have multiple items with the exact same value.
4. **Indexed:** Each item has an index starting from `0`.

#### **Common List Methods (Attributes):**
* `append(x)`: Adds an item `x` to the end of the list.
* `insert(index, x)`: Inserts an item `x` at a specific position.
* `remove(x)`: Removes the first occurrence of item `x`.
* `pop(index)`: Removes and returns the item at the given index (defaults to the last item).
* `sort()`: Sorts the list in ascending order.
* `reverse()`: Reverses the order of the list elements.

````python
# 1. Creating a list
fruits = ["apple", "banana", "cherry"]
print("Original list:", fruits)

# 2. Modifying an item by index
fruits[1] = "mango"
print("After modification:", fruits)

# 3. Appending and Inserting items
fruits.append("orange")           # Adds to the end
fruits.insert(1, "strawberry")    # Inserts at index 1
print("After adding elements:", fruits)

# 4. Removing items
fruits.remove("cherry")           # Removes 'cherry'
popped_item = fruits.pop(0)       # Removes and returns index 0
print(f"Removed item: {popped_item}")
print("After removal steps:", fruits)

# 5. Sorting and Reversing
fruits.sort()
print("Sorted list:", fruits)
fruits.reverse()
print("Reversed list:", fruits)
````

### **9. Tuples in Python**

A **Tuple** is an ordered, unchangeable (immutable) collection of items. It is written inside round parentheses `()`.

#### **Key Attributes & Characteristics:**
1. **Ordered:** Items have a fixed, defined order.
2. **Immutable (Unchangeable):** You cannot add, remove, or modify items once the tuple is created. This makes tuples faster than lists and safe from accidental changes.
3. **Allows Duplicates:** You can have duplicate values.
4. **Indexed:** You can access items using index numbers.

#### **Common Tuple Methods (Attributes):**
* `count(x)`: Returns the total number of times a value `x` appears in the tuple.
* `index(x)`: Searches for a value `x` and returns its first index position.

````python
# 1. Creating a tuple
colors = ("red", "green", "blue", "red")
print("Tuple items:", colors)

# 2. Accessing items using indexing
print("First color:", colors[0])
print("Last color:", colors[-1])

# 3. Tuple attributes/methods
red_count = colors.count("red")
print("How many times 'red' appears?:", red_count)

green_index = colors.index("green")
print("Index of 'green':", green_index)

# Note: Trying to do 'colors[0] = "yellow"' will raise a TypeError because tuples are immutable!
````

### **10. Sets in Python**

A **Set** is an unordered collection of unique elements written inside curly braces `{}`.

#### **Key Attributes & Characteristics:**
1. **Unordered:** Items do not have a defined order, and they can appear in different orders every time you run the code.
2. **Unindexed:** You cannot access items using index positions like `set[0]`.
3. **Unique Items Only:** Sets automatically remove any duplicate values.
4. **Mutable:** You can add or remove items, but the individual items inside must be immutable.

#### **Common Set Methods (Attributes):**
* `add(x)`: Adds element `x` to the set.
* `remove(x)` / `discard(x)`: Removes element `x`. (`remove` raises an error if the element doesn't exist, while `discard` does not).
* `union(other_set)`: Combines elements of both sets.
* `intersection(other_set)`: Returns only the elements present in both sets.
* `difference(other_set)`: Returns elements present in the first set but not in the second.

````python
# 1. Creating a set with duplicate values
numbers = {1, 2, 2, 3, 4, 4}
print("Set removes duplicates automatically:", numbers)

# 2. Adding and Removing elements
numbers.add(5)
numbers.discard(2)   # Safely removes 2
print("After adding 5 and discarding 2:", numbers)

# 3. Set operations (Attributes)
set_a = {1, 2, 3}
set_b = {3, 4, 5}

# Union (all elements)
print("Union:", set_a.union(set_b))

# Intersection (common elements)
print("Intersection:", set_a.intersection(set_b))

# Difference (elements in set_a but not in set_b)
print("Difference (A - B):", set_a.difference(set_b))
````

### **11. Dictionaries in Python**

A **Dictionary** stores data as Key-Value pairs, written inside curly braces `{key: value}`.

#### **Key Attributes & Characteristics:**
1. **Key-Value Pairs:** Each entry has an identifying key and a corresponding value.
2. **Ordered:** Since Python 3.7, dictionaries maintain insertion order.
3. **Mutable:** You can add, change, or delete keys and values.
4. **Unique Keys:** Duplicate keys are not allowed. If you try to assign a value to an existing key, it overwrites the old value.

#### **Common Dictionary Methods (Attributes):**
* `keys()`: Returns a list-like view of all keys in the dictionary.
* `values()`: Returns a list-like view of all values in the dictionary.
* `items()`: Returns a list of key-value tuples.
* `get(key)`: Returns the value of a key safely (does not raise an error if the key doesn't exist).
* `update({key: value})`: Adds or updates key-value pairs.
* `pop(key)`: Removes the specified key and returns its value.

````python
# 1. Creating a dictionary
student_info = {
    "name": "Sanjana",
    "course": "Python",
    "score": 95
}
print("Original Dictionary:", student_info)

# 2. Accessing values safely using get()
print("Student Name:", student_info.get("name"))
print("Age (Key not present):", student_info.get("age", "Not Specified"))

# 3. Modifying and updating values
student_info["score"] = 98
student_info.update({"grade": "A+"})
print("After Update:", student_info)

# 4. Getting Keys, Values, and Items
print("Keys list:", list(student_info.keys()))
print("Values list:", list(student_info.values()))
print("Items pairs:", list(student_info.items()))

# 5. Removing a key
removed_value = student_info.pop("course")
print(f"Removed Course: {removed_value}")
print("Final Dictionary:", student_info)
````

## **12. Exception Handling**

Sometimes, errors occur during code execution that cause the program to crash. **Exception Handling** helps us deal with these errors gracefully without stopping the entire program.
* **`try`**: This block lets you test a block of code for errors.
* **`except`**: This block lets you handle the error if one occurs in the try block.

````python
try:
    number = int(input("Enter a number to divide 10: "))
    result = 10 / number
    print("Result is:", result)
except ZeroDivisionError:
    print("Error: You cannot divide a number by zero!")
except ValueError:
    print("Error: Please enter a valid integer number!")
````

## **14. Object-Oriented Programming (OOPs)**

**OOPs** is a programming style that allows us to model real-world things using code.
* **Class**: A blueprint or template for creating objects (e.g., a blueprint for a house).
* **Object**: The actual instance created from a class (e.g., the physical house built from the blueprint).
* **`__init__` (Constructor)**: A special method that runs automatically whenever we create a new object.

````python
class Dog:
    # Constructor method
    def __init__(self, name, breed):
        self.name = name
        self.breed = breed

    def bark(self):
        return f"{self.name} says Woof!"

# Creating objects
my_dog = Dog("Buddy", "Golden Retriever")
print("Dog Name:", my_dog.name)
print(my_dog.bark())
````

## **16. OOPs Project: Bank Management System**

Let's build a simple bank account management program. We will use a `BankAccount` class to keep track of an owner's name and balance, allowing them to deposit money, withdraw money, and view their balance.

````python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print(f"Rupees {amount} deposited. Current Balance: {self.balance}")

    def withdraw(self, amount):
        if amount > self.balance:
            print("Insufficient funds!")
        else:
            self.balance -= amount
            print(f"Rupees {amount} withdrawn. Remaining Balance: {self.balance}")

    def check_balance(self):
        print(f"Account Owner: {self.owner} | Available Balance: {self.balance}")

# Testing Bank Project
account = BankAccount("Sanjana", 500)
account.check_balance()
account.deposit(200)
account.withdraw(150)
account.check_balance()
````
